```python
# conv VAE-GMM: same model as seperate_benchmark/vae_train.ipynb but with a CONV encoder/decoder
# (Conv1d + dilated Conv1d) in place of the MLP. Everything downstream of z -- the variational-Bayes
# GMM (Dirichlet-pi posterior + Normal-Gamma components, closed-form M-steps) and the shared
# background Gaussian -- is unchanged.
import sys, os, numpy as np, pandas as pd
import h5py
import matplotlib.pyplot as plt
import torch, torch.nn as nn, torch.nn.functional as F

seam_dir = "/grid/koo/home/pmantill/projects/SEAM_validate/results/single_seq"
peak = "peak84030"                        # top K562 lentiMPRA seed (seam_seeds.csv)
lib    = "100k"                           # library size: "100k" or "200k"
suffix = {"100k": "", "200k": "_200k"}[lib]

with h5py.File(f"{seam_dir}/ism{suffix}/{peak}.h5") as f:
    attributions = f["ism_centered"][:]   # (N,200,4) ISM, mean-centered across channels
    predictions  = f["predictions"][:]    # (N,) self-prediction of each mutant

L = attributions.shape[1]
X = attributions.reshape(attributions.shape[0], -1).astype(np.float32)   # (N,4L) position-major
# per-map normalization to unit STANDARD DEVIATION: removes each map's overall height/magnitude so the
# GMM clusters on MECHANISM (shape), while keeping the data at O(1) scale. (L2 unit-norm removes the same
# magnitude but shrinks the map to ~1/sqrt(D) -> background-Gaussian precision ~1300 -> log_var_b collapse
# -> NaN. unit-std is the same scale-invariance at a scale the Gaussian likelihood is stable at, prec ~1.7.)
X /= (X.std(axis=1, keepdims=True) + 1e-8)
N, D = X.shape
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
X_t = torch.from_numpy(X).to(device)
var_data = X.var(0) + 1e-8                 # (D,) global per-dim variance (bg cov seed)
print(f"[{lib}] N={N:,}  D={D} (=4*{L})  device={device}  (per-map unit-std normalized)")
```

```python
# ----- hyperparameters (mixture side unchanged: K=200, alpha0=1e-3) -----
J        = 8         # latent (foreground code) dimension
K        = 200       # initial number of GMM components (sparse Dirichlet prunes unused)
alpha0   = 1e-3      # Dirichlet concentration (alpha0 < 1 -> sparse -> learns K)
Cch      = 32        # conv channels
kernel   = 7
dilation = 4         # dilation of the second conv (widens receptive field w/o extra layers)
seed     = 4
torch.manual_seed(seed); rng = np.random.default_rng(seed)


class Exp(nn.Module):
    """exponential activation (Koo & Ploenzke 2021 motif layer). BatchNorm precedes it so the
    pre-activation is ~unit-scale and exp does not overflow to inf."""
    def forward(self, x):
        return torch.exp(x)


class VAE_GMM(nn.Module):
    """Conv-encoder/decoder VAE over the foreground f=decoder(z); shared bg Gaussian N(mu_b,sigma_b^2)
    added before the reconstruction; GMM(K) prior on z. Mixture weights: Dirichlet(alpha0) posterior.
    Component (mean, precision): diagonal Normal-Gamma prior + variational NG posterior, both updated
    by closed-form M-steps. Identical to the MLP VAE_GMM except enc/dec are convolutional."""
    def __init__(self, D, L, J, K, alpha0, Cch, kernel, dilation, var_data,
                 m0=0.0, kappa0=1.0, a0=1.0, b0=1.0):        # weakly-informative NG prior (tunable)
        super().__init__()
        self.D, self.L, self.J, self.K, self.alpha0, self.Cch = D, L, J, K, alpha0, Cch
        pad_d = dilation * (kernel - 1) // 2                       # 'same' padding for the dilated conv
        # encoder q(z|x): motif layer Conv1d->BatchNorm->exp, then dilated Conv1d->ReLU -> flatten -> J
        # (exp only on the first conv, per Koo & Ploenzke; BatchNorm keeps its pre-activation from overflowing)
        self.enc = nn.Sequential(
            nn.Conv1d(4, Cch, kernel, padding=kernel // 2), nn.BatchNorm1d(Cch), Exp(),
            nn.Conv1d(Cch, Cch, kernel, padding=pad_d, dilation=dilation), nn.ReLU())
        self.enc_mu = nn.Linear(Cch * L, J)
        self.enc_lv = nn.Linear(Cch * L, J)
        # decoder p(f|z): J -> (Cch,L) -> dilated Conv1d->ReLU -> Conv1d -> (4,L) foreground map
        self.dec_fc = nn.Linear(J, Cch * L)
        self.dec = nn.Sequential(
            nn.Conv1d(Cch, Cch, kernel, padding=pad_d, dilation=dilation), nn.ReLU(),
            nn.Conv1d(Cch, 4, kernel, padding=kernel // 2))
        # shared background Gaussian: free mean starts at ZERO; cov is ISOTROPIC (scalar)
        self.mu_b      = nn.Parameter(torch.zeros(D))
        self.log_var_b = nn.Parameter(torch.tensor(float(np.log(var_data.mean()))))
        # ---- GMM components: Normal-Gamma prior + variational NG posterior (BUFFERS, set by M-steps) ----
        self.m0, self.kappa0, self.a0, self.b0 = float(m0), float(kappa0), float(a0), float(b0)
        self.register_buffer("m_",     torch.zeros(K, J))
        self.register_buffer("kappa_", torch.full((K,), float(kappa0)))
        self.register_buffer("a_",     torch.full((K,), float(a0)))
        self.register_buffer("b_",     torch.full((K, J), float(b0)))
        self.register_buffer("alpha_", torch.full((K,), alpha0 + 1.0))    # q(pi)=Dir(alpha), M-step buffer

    @property
    def alpha(self):
        return self.alpha_

    def ng_expect(self):
        """expectations under q(mu,tau)=NG: E[tau]=a/b, E[log tau]=psi(a)-log b  -> both (K,J)."""
        E_tau    = self.a_[:, None] / self.b_
        E_logtau = torch.digamma(self.a_)[:, None] - torch.log(self.b_)
        return E_tau, E_logtau

    def elogN(self, mu_z, s2):
        """E_{q(mu,tau)}[log N(z|mu_k,tau_k^-1)] averaged over q(z|x)=N(mu_z,s2) -> (B,K)."""
        E_tau, E_logtau = self.ng_expect()                              # (K,J)
        diff2 = (mu_z[:, None, :] - self.m_[None])**2 + s2[:, None, :]  # (B,K,J)
        return 0.5 * ((E_logtau[None] - np.log(2*np.pi))
                      - E_tau[None] * diff2
                      - (1.0 / self.kappa_)[None, :, None]).sum(-1)

    def kl_ng(self):
        """sum_{k,j} KL[ NG_q(m,kappa,a,b) || NG_prior(m0,kappa0,a0,b0) ]  (scalar)."""
        aE, b, k, m = self.a_[:, None], self.b_, self.kappa_, self.m_
        a0, b0, k0, m0 = self.a0, self.b0, self.kappa0, self.m0
        kl_gamma = (a0 * torch.log(b / b0) + (aE - a0) * torch.digamma(aE)
                    - torch.lgamma(aE) + torch.lgamma(torch.tensor(a0)) + aE * (b0 - b) / b)
        kl_gauss = (0.5 * torch.log(k / k0)[:, None] + 0.5 * (k0 / k)[:, None] - 0.5
                    + 0.5 * k0 * (aE / b) * (m - m0)**2)
        return (kl_gamma + kl_gauss).sum()

    @torch.no_grad()
    def m_step_alpha(self, Nk):
        """Exact ELBO maximizer for the Dirichlet posterior: alpha_k = alpha0 + N_k."""
        self.alpha_.copy_(self.alpha0 + torch.as_tensor(Nk, dtype=torch.float32, device=self.alpha_.device))

    @torch.no_grad()
    def m_step_ng(self, Nk, S1, S2):
        """Closed-form CAVI update of the component NG posterior (Bishop 10.60-10.63, diagonal)."""
        Nk_safe = Nk.clamp_min(1e-8)
        self.kappa_.copy_(self.kappa0 + Nk)
        self.a_.copy_(self.a0 + 0.5 * Nk)
        self.m_.copy_((self.kappa0 * self.m0 + S1) / self.kappa_[:, None])
        xbar = S1 / Nk_safe[:, None]
        NkS  = S2 - S1**2 / Nk_safe[:, None]
        coef = (self.kappa0 * Nk / (self.kappa0 + Nk))[:, None]
        b_new = self.b0 + 0.5 * (NkS + coef * (xbar - self.m0)**2)
        self.b_.copy_(b_new.clamp_min(1e-8))

    def encode(self, x):
        h = self.enc(x.view(-1, self.L, 4).transpose(1, 2))     # (B,4,L) channels-first for Conv1d
        h = h.reshape(h.shape[0], -1)
        return self.enc_mu(h), self.enc_lv(h)

    def decode(self, z):
        h = self.dec(self.dec_fc(z).view(-1, self.Cch, self.L))  # (B,4,L)
        return h.transpose(1, 2).reshape(z.shape[0], -1)         # (B,D) back in x's position-major layout

    def recon_loss(self, x):
        # STACKED-AUTOENCODER warmup: reconstruction (MSE) ONLY -- deterministic z=mu_z, no reparam, no KL
        mu_z, _ = self.encode(x)
        recon = self.decode(mu_z) + self.mu_b
        return ((x - recon)**2).sum(-1).mean()

    def forward(self, x, n_total):
        mu_z, lv_z = self.encode(x)                              # (B,J)
        s2 = torch.exp(lv_z)
        z = mu_z + torch.randn_like(mu_z) * torch.sqrt(s2)
        recon = self.decode(z) + self.mu_b                       # f + b  (mean of x)

        log_p_x = -0.5 * (np.log(2*np.pi) + self.log_var_b
                          + (x - recon)**2 * torch.exp(-self.log_var_b)).sum(-1)   # (B,)

        ElogN   = self.elogN(mu_z, s2)                               # (B,K)
        Elog_pi = torch.digamma(self.alpha) - torch.digamma(self.alpha.sum())   # (K,)
        log_un  = Elog_pi[None] + ElogN                             # (B,K)
        log_gamma = log_un - torch.logsumexp(log_un, 1, keepdim=True)
        gamma = log_gamma.exp().detach()                             # responsibilities (E-step)

        Hqz      = 0.5 * (1.0 + np.log(2*np.pi) + lv_z).sum(-1)      # entropy of q(z|x)
        klz_term = (gamma * ElogN).sum(-1) + Hqz
        cat_term = (gamma * (Elog_pi[None] - log_gamma)).sum(-1)
        elbo_n   = log_p_x + klz_term + cat_term                    # (B,)

        alpha = self.alpha
        kl_pi = (torch.lgamma(alpha.sum()) - torch.lgamma(alpha).sum()
                 - torch.lgamma(torch.tensor(self.K * alpha0, device=x.device))
                 + self.K * torch.lgamma(torch.tensor(alpha0, device=x.device))
                 + ((alpha - alpha0) * Elog_pi).sum())

        loss = -elbo_n.mean() + (kl_pi + self.kl_ng()) / n_total
        return loss, elbo_n.mean().detach(), gamma, mu_z.detach(), s2.detach()

model = VAE_GMM(D, L, J, K, alpha0, Cch, kernel, dilation, var_data).to(device)
print(model)
```

```python
# ----- WARMUP: stacked autoencoder -- train encoder+decoder (+bg mean) on RECONSTRUCTION (MSE) ONLY -----
# deterministic z=mu_z, no reparam, no KL. Just makes the latent meaningful before clustering.
warmup_epochs, warmup_batch, warmup_lr = 30, 256, 1e-3

warm_params = list(model.enc.parameters()) + list(model.enc_mu.parameters()) \
            + list(model.dec_fc.parameters()) + list(model.dec.parameters()) + [model.mu_b]
opt_w = torch.optim.Adam(warm_params, lr=warmup_lr)

model.train()
warm_losses = []
for ep in range(warmup_epochs):
    perm = torch.randperm(N, device=device)
    tot = 0.0
    for i in range(0, N, warmup_batch):
        xb = X_t[perm[i:i+warmup_batch]]
        opt_w.zero_grad()
        l = model.recon_loss(xb)
        l.backward(); opt_w.step()
        tot += l.item() * xb.shape[0]
    warm_losses.append(tot / N)
    if ep % 5 == 0 or ep == warmup_epochs - 1:
        print(f"warmup epoch {ep:3d}  recon MSE/n {warm_losses[-1]:.5f}")

plt.plot(warm_losses); plt.xlabel("warmup epoch"); plt.ylabel("recon MSE / n"); plt.title("stacked-AE warmup"); plt.show()
```

```python
# ----- init the component NG posterior: means at K random DATA latents, width = per-dim latent scale -----
#   m_     <- K randomly drawn data-point latent codes (every component starts ON data)
#   kappa_ <- kappa0, a_ <- a0
#   b_     <- a0 * Var_j[Z] per latent dim  ->  E[tau_j]=a0/b_j=1/var_j  ->  sigma_j = latent std_j
# The closed-form M-step then tightens each component from the responsibility-weighted z statistics.
model.eval()
with torch.no_grad():
    Z = torch.cat([model.encode(X_t[i:i+4096])[0] for i in range(0, N, 4096)]).cpu().numpy()  # (N,J)

Zvar = Z.var(0) + 1e-8                                        # (J,) per-dim latent variance
idx0 = rng.choice(N, K, replace=False)                        # K distinct data-point latent codes
with torch.no_grad():
    model.m_.copy_(torch.from_numpy(Z[idx0]).to(device))
    model.kappa_.fill_(model.kappa0)
    model.a_.fill_(model.a0)
    model.b_.copy_(torch.from_numpy((model.a0 * Zvar).astype(np.float32)).to(device))   # (J,)->(K,J)

spread = Z.std(0).mean()
E_tau  = (model.a_[:, None] / model.b_).cpu().numpy()
sig    = float(np.mean(1.0 / np.sqrt(E_tau)))
sep    = model.m_.std().item()
print(f"m_=K random data latents, b_=per-dim var | latent spread {spread:.3f} | m_ std={sep:.3f} | "
      f"sigma={sig:.3f} -> sigma/spread {sig/spread:.2f} | separation/width {sep/sig:.3f}")
```

```python
# ----- train full ELBO (Adam for encoder/decoder/bg) + closed-form M-steps each epoch -----
#   - alpha (Dirichlet) and the component NG posterior (m,kappa,a,b) are buffers -> excluded from Adam
#   - each epoch accumulates N_k = sum_n gamma_nk, S1 = sum_n gamma*mu_z, S2 = sum_n gamma*(mu_z^2+s2)
#   - epoch end: alpha_k <- alpha0 + N_k  AND  NG posterior <- CAVI update (exact ELBO maximizers)
out_dir = "/grid/koo/home/pmantill/projects/VBI/VAE_gmm/conv_VAE/results"
os.makedirs(out_dir, exist_ok=True)
# tag encodes the preprocessing + activation (l2norm input, BatchNorm+exp motif layer) so this run does
# NOT collide with / load the older raw-input, all-exp checkpoints (different architecture -> different keys).
hp_tag   = f"{peak}.{lib}.conv.J{J}.K{K}.c{Cch}.k{kernel}.d{dilation}.a{alpha0:g}.warm{warmup_epochs}.l2norm.bnexp.mstep.ng.rfdmu.seed{seed}"
ckpt_path = f"{out_dir}/{hp_tag}.pt"

epochs, batch, lr = 100, 256, 1e-5

if os.path.exists(ckpt_path):
    ck = torch.load(ckpt_path, map_location=device, weights_only=False)
    model.load_state_dict(ck["state_dict"]); elbos = ck["elbos"]; n_alive_hist = ck["n_alive_hist"]
    alpha_hist = ck["alpha_hist"]
    print(f"loaded cached fit: {ckpt_path}  ({len(elbos)} epochs, final ELBO {elbos[-1]:.3f})")
else:
    # snapshot the post-warmup / pre-VB state so the benchmark can isolate what the VB M-steps add.
    warm_ckpt = f"{out_dir}/{hp_tag}.warmup_only.pt"
    torch.save({"state_dict": model.state_dict(),
                "hp": dict(peak=peak, lib=lib, J=J, K=K, Cch=Cch, kernel=kernel, dilation=dilation,
                           alpha0=alpha0, warmup_epochs=warmup_epochs, seed=seed, prior="normal_gamma",
                           stage="warmup_only", m0=model.m0, kappa0=model.kappa0,
                           a0=model.a0, b0=model.b0)}, warm_ckpt)
    print(f"saved warmup-only snapshot -> {warm_ckpt}")

    opt = torch.optim.Adam(model.parameters(), lr=lr)     # buffers (alpha, NG posterior) not in parameters()
    elbos, n_alive_hist, alpha_hist = [], [], []
    for ep in range(epochs):
        perm = torch.randperm(N, device=device)
        ep_elbo = 0.0
        Nk = torch.zeros(K, device=device)
        S1 = torch.zeros(K, J, device=device)
        S2 = torch.zeros(K, J, device=device)
        won = torch.zeros(K, dtype=torch.bool, device=device)   # component won >=1 argmax this epoch
        model.train()
        for i in range(0, N, batch):
            xb = X_t[perm[i:i+batch]]
            opt.zero_grad()
            loss, elbo, gamma, mu_z, s2 = model(xb, N)
            loss.backward(); opt.step()
            ep_elbo += float(elbo) * xb.shape[0]
            Nk += gamma.sum(0)                       # gamma, mu_z, s2 already detached in forward
            S1 += gamma.t() @ mu_z                   # (K,J)
            S2 += gamma.t() @ (mu_z**2 + s2)         # (K,J)
            won[gamma.argmax(1)] = True
        model.m_step_alpha(Nk)                       # exact M-step on the Dirichlet posterior
        model.m_step_ng(Nk, S1, S2)                  # exact CAVI M-step on the component NG posterior
        elbos.append(ep_elbo / N)
        assert np.isfinite(elbos[-1]), f"training diverged at epoch {ep}: ELBO/n={elbos[-1]} (exp overflow?)"
        n_alive_hist.append(int(won.sum()))
        with torch.no_grad():
            a = model.alpha.cpu().numpy()
        alpha_hist.append([float(a.min()), float(np.median(a)), float(a.max())])
        if ep % 10 == 0 or ep == epochs - 1:
            print(f"epoch {ep:3d}  ELBO/n {elbos[-1]:8.3f}  surviving(argmax) {n_alive_hist[-1]:3d}"
                  f"  alpha [{a.min():.3f}, {np.median(a):.3f}, {a.max():.0f}]")
    torch.save({"state_dict": model.state_dict(), "elbos": elbos, "n_alive_hist": n_alive_hist,
                "alpha_hist": alpha_hist,
                "hp": dict(peak=peak, lib=lib, J=J, K=K, Cch=Cch, kernel=kernel, dilation=dilation,
                           alpha0=alpha0, warmup_epochs=warmup_epochs, seed=seed, mstep=True,
                           prior="normal_gamma", m0=model.m0, kappa0=model.kappa0,
                           a0=model.a0, b0=model.b0, init="random_codes_ng_prior")}, ckpt_path)
    print(f"saved -> {ckpt_path}")

ah = np.array(alpha_hist)
fig, ax = plt.subplots(1, 3, figsize=(17, 3))
ax[0].plot(elbos);        ax[0].set_xlabel("epoch"); ax[0].set_ylabel("ELBO / n")
ax[1].plot(n_alive_hist); ax[1].set_xlabel("epoch"); ax[1].set_ylabel("# surviving components (argmax)")
ax[2].semilogy(ah[:, 2], label="max"); ax[2].semilogy(ah[:, 1], label="median"); ax[2].semilogy(ah[:, 0], label="min")
ax[2].set_xlabel("epoch"); ax[2].set_ylabel("alpha"); ax[2].legend(fontsize=7)
ax[2].set_title("Dirichlet posterior after M-step")
plt.tight_layout(); plt.show()
```

```python
# ----- CSM: cluster summary matrix, within-cluster positional entropy -----
# Hard-assign every ISM variant to its GMM cluster (NG E-step: argmax over Elog_pi + E_q[log N(z|mu_k)]),
# then per cluster take the per-position Shannon entropy (bits) of the member one-hots.
# Low entropy = position the cluster holds fixed (conserved motif); high = free to vary.
ADAPTER = 15   # 15bp adapter each side (context for ISM, not attributed / not in the 800-D input)

model.eval()
Elog_pi = torch.digamma(model.alpha) - torch.digamma(model.alpha.sum())
hard = []
with torch.no_grad():
    for i in range(0, N, 4096):
        mu_z, lv_z = model.encode(X_t[i:i+4096])
        ElogN = model.elogN(mu_z, torch.exp(lv_z))
        hard.append((Elog_pi[None] + ElogN).argmax(1).cpu())
hard = torch.cat(hard).numpy()

with h5py.File(f"{seam_dir}/mutagenesis_lib{suffix}/{peak}.h5") as f:
    onehot = f["sequences"][:][:, ADAPTER:-ADAPTER]        # (N,200,4) core, aligned to attribution positions

alive = sorted(np.unique(hard), key=lambda k: (hard == k).sum(), reverse=True)   # largest first
entropy = np.zeros((len(alive), L))
for i, k in enumerate(alive):
    p = onehot[hard == k].mean(0)                          # (L,4) base frequencies
    entropy[i] = -(p * np.log2(p + 1e-12)).sum(1)          # (L,) bits, 0..2
print(f"{hp_tag}\n{len(alive)} clusters x {L} positions")

plt.figure(figsize=(14, max(4, 0.22 * len(alive))))
plt.imshow(entropy, aspect="auto", cmap="Reds", vmin=0, vmax=2, interpolation="nearest")
plt.colorbar(label="within-cluster entropy (bits)")
plt.yticks(range(len(alive)), [f"cl {k} (n={(hard == k).sum()})" for k in alive], fontsize=6)
plt.xlabel("position"); plt.ylabel("cluster (largest first)")
plt.title(f"conv VAE-GMM CSM — within-cluster positional entropy  |  {hp_tag}", fontsize=8)
plt.tight_layout(); plt.show()
```

```python
# ----- load the seperate_benchmark MLP VAE-GMM checkpoints (same peak+lib) for side-by-side comparison -----
# IMPORTANT: these MLP checkpoints were trained on RAW (un-normalized) attribution maps, so they must be
# scored on RAW input -- NOT the per-map L2-normalized X_t the conv model now uses. Feeding a model input
# preprocessed differently than it was trained on gives out-of-distribution, degenerate assignments.
import glob
X_raw   = attributions.reshape(attributions.shape[0], -1).astype(np.float32)   # raw maps, no normalization
X_raw_t = torch.from_numpy(X_raw).to(device)

def ng_hard_labels(sd, Xin):
    """NG hard cluster labels from an MLP checkpoint applied to Xin. None for unrecognized architectures."""
    sd = {k: v.to(device) for k, v in sd.items()}
    if   "enc_in.weight" in sd: arch = "skip"
    elif "enc.0.weight"  in sd: arch = "plain"
    else: return None
    alpha = sd["alpha_"]; Elog_pi = torch.digamma(alpha) - torch.digamma(alpha.sum())
    E_tau    = sd["a_"][:, None] / sd["b_"]
    E_logtau = torch.digamma(sd["a_"])[:, None] - torch.log(sd["b_"])
    out = []
    with torch.no_grad():
        for i in range(0, N, 4096):
            x = Xin[i:i+4096]
            if arch == "skip":
                h = F.relu(x @ sd["enc_in.weight"].t() + sd["enc_in.bias"])
                h = h + F.relu(h @ sd["enc_res.weight"].t() + sd["enc_res.bias"])
            else:
                h = F.relu(x @ sd["enc.0.weight"].t() + sd["enc.0.bias"])
            mu_z = h @ sd["enc_mu.weight"].t() + sd["enc_mu.bias"]
            s2   = torch.exp(h @ sd["enc_lv.weight"].t() + sd["enc_lv.bias"])
            diff2 = (mu_z[:, None, :] - sd["m_"][None])**2 + s2[:, None, :]
            score = 0.5 * ((E_logtau[None] - np.log(2*np.pi)) - E_tau[None] * diff2
                           - (1.0 / sd["kappa_"])[None, :, None]).sum(-1)
            out.append((Elog_pi[None] + score).argmax(1).cpu())
    return torch.cat(out).numpy()

BENCH_DIR = "/grid/koo/home/pmantill/projects/VBI/VAE_gmm/seperate_benchmark/results"
cps = [cp for cp in sorted(glob.glob(f"{BENCH_DIR}/{peak}.{lib}.*.pt")) if "warmup_only" not in cp]
cols = plt.cm.turbo(np.linspace(0.05, 0.95, max(len(cps), 1)))   # unique color per config, no recycling
bench_ckpts = []
for cp, col in zip(cps, cols):
    lab = ng_hard_labels(torch.load(cp, map_location=device, weights_only=False)["state_dict"], X_raw_t)
    if lab is None:
        print(f"skip {os.path.basename(cp)[:-3]}: unrecognized architecture"); continue
    tag = os.path.basename(cp)[:-3].replace(f"{peak}.{lib}.", "")   # short config label
    bench_ckpts.append((tag, lab, tuple(col)))
    print(f"{tag}  surviving={len(np.unique(lab))}")
print(f"{len(bench_ckpts)} seperate_benchmark configs loaded (scored on RAW input)")
```

```python
# ----- benchmark 1: functional coherence (eta^2 of activity ~ cluster) -----
# Hard-assign each ISM map to a cluster, then measure the fraction of variance in the self-prediction
# explained by the cluster label. 1 = each cluster is functionally homogeneous. Compares our conv
# VAE-GMM (labels from the live model, `hard`) against the SEAM kmeans + random baselines AND every
# seperate_benchmark MLP config (bench_ckpts). Bar label = benchmark value + surviving-cluster count.
def eta2(labels, y):
    """one-way ANOVA eta^2: between-group variance of y / total variance of y."""
    ybar = y.mean(); ss_tot = ((y - ybar)**2).sum()
    ss_bet = sum(len(y[labels == g]) * (y[labels == g].mean() - ybar)**2 for g in np.unique(labels))
    return ss_bet / ss_tot

seam_k = 100
seam_labels = np.load(f"{seam_dir}/foregrounds{suffix}_kmeans{seam_k}/{peak}/cluster_labels.npy")
assert len(seam_labels) == N, f"SEAM labels {len(seam_labels)} != N {N} (library mismatch)"
rand_labels = np.random.default_rng(0).integers(0, 100, N)

label_sets = [(f"SEAM.kmeans{seam_k}", seam_labels, "#DD8452"),
              ("RANDOM.k100",          rand_labels, "#C44E52"),
              (f"conv VAE-GMM (J{J})", hard,        "#4C72B0")] + bench_ckpts
names   = [nm for nm, _, _ in label_sets]
colors  = [c  for _, _, c in label_sets]
survivors = [len(np.unique(lab)) for _, lab, _ in label_sets]

coherences = [eta2(lab, predictions) for _, lab, _ in label_sets]
for nm, coh, s in zip(names, coherences, survivors):
    print(f"{nm}  surviving clusters={s:3d}  functional coherence (eta^2)={coh:.3f}")

fig, ax = plt.subplots(figsize=(0.6 * len(names) + 3, 5))
ax.bar(range(len(names)), coherences, color=colors)
for i, (coh, s) in enumerate(zip(coherences, survivors)):
    ax.text(i, coh + 0.01, f"{coh:.3f}\n{s} cl", ha="center", va="bottom", fontsize=7)
ax.set_xticks(range(len(names))); ax.set_xticklabels(names, rotation=45, ha="right", fontsize=7)
ax.set_ylim(0, 1.1); ax.set_ylabel(r"functional coherence  ($\eta^2$ of activity ~ cluster)")
plt.tight_layout(); plt.show()
```

```python
# ----- benchmark 2: seqlet-overlap coherence -----
# Recursive-seqlet-call every attribution map (tangermeme), then score each clustering by how much the
# seqlets of same-cluster sequences occupy the SAME positions: per sequence, the leave-one-out fraction
# of its seqlet positions its cluster-mates also cover. 1 = perfect within-cluster alignment.
from tangermeme.seqlet import recursive_seqlets

occ_cache = f"{out_dir}/{peak}.{lib}.seqlet_occupancy.npz"
shared    = f"/grid/koo/home/pmantill/projects/VBI/VAE_gmm/seperate_benchmark/results/{peak}.{lib}.seqlet_occupancy.npz"
src = occ_cache if os.path.exists(occ_cache) else (shared if os.path.exists(shared) else None)
if src:
    occ = np.unpackbits(np.load(src)["occ"], axis=1)[:, :L].astype(bool)
    print(f"loaded seqlet occupancy: {src}  ({occ.any(1).sum():,} maps with >=1 seqlet)")
else:
    with h5py.File(f"{seam_dir}/mutagenesis_lib{suffix}/{peak}.h5") as f:
        seqs = f["sequences"][:, 15:215, :]                    # (N,L,4) aligned to attributions
    proj = (attributions * seqs).sum(-1)                       # (N,L) projected attribution of observed base
    df = recursive_seqlets(proj)                               # positive seqlets: example_idx,start,end,...
    occ = np.zeros((N, L), dtype=bool)
    for e, a, b in zip(df["example_idx"].to_numpy(), df["start"].to_numpy(), df["end"].to_numpy()):
        occ[e, a:b] = True
    np.savez_compressed(occ_cache, occ=np.packbits(occ, axis=1))
    print(f"called {len(df):,} seqlets; {occ.any(1).sum():,} maps with >=1 -> {occ_cache}")

def seqlet_overlap(labels):
    """mean leave-one-out fraction of each seq's seqlet positions shared by its cluster-mates."""
    labels = np.asarray(labels); Kc = int(labels.max()) + 1
    C = np.zeros((Kc, L)); np.add.at(C, labels, occ)          # (Kc,L) per-cluster coverage
    d  = occ.sum(1)                                            # (N,) |seqlet positions| per seq
    oc = (occ * C[labels]).sum(1)                              # (N,) cluster counts over its positions
    sizes = np.bincount(labels, minlength=Kc); nk = sizes[labels]
    valid = (d > 0) & (nk > 1)
    score = (oc[valid] - d[valid]) / ((nk[valid] - 1) * d[valid])
    return float(score.mean()), int(valid.sum())

overlaps = [seqlet_overlap(lab)[0] for _, lab, _ in label_sets]
for nm, o, s in zip(names, overlaps, survivors):
    print(f"{nm}  seqlet-overlap={o:.3f}  ({s} clusters)")

fig, ax = plt.subplots(figsize=(0.6 * len(names) + 3, 5))
ax.bar(range(len(names)), overlaps, color=colors)
for i, (o, s) in enumerate(zip(overlaps, survivors)):
    ax.text(i, o + 0.01, f"{o:.3f}\n{s} cl", ha="center", va="bottom", fontsize=7)
ax.set_xticks(range(len(names))); ax.set_xticklabels(names, rotation=45, ha="right", fontsize=7)
ax.set_ylim(0, 1.1); ax.set_ylabel("seqlet-overlap coherence  (leave-one-out cluster-mate agreement)")
plt.tight_layout(); plt.show()
```

```python
# ----- benchmark 3: seqlet-overlap coherence NORMALIZED by across-cluster overlap -----
# Benchmark 2's raw overlap is inflated by globally popular positions (the WT motif), so even random
# clusters score high. Here we divide each seq's WITHIN-cluster overlap by its ACROSS-cluster overlap:
#   ratio_i = within_i / max(across_i, p0)   ->  1 = chance,  >1 = concentration  (p0 = global chance rate)
# Cluster-exclusive seqlets (across_i=0) are KEPT: across is floored at p0 so exclusives score high.
def seqlet_overlap_norm(labels):
    labels = np.asarray(labels); Kc = int(labels.max()) + 1
    C = np.zeros((Kc, L)); np.add.at(C, labels, occ); Ctot = occ.sum(0)   # per-cluster + global coverage
    d  = occ.sum(1); Ck = C[labels]
    oc_in  = (occ * Ck).sum(1)                                # own-cluster count over its positions
    oc_tot = (occ * Ctot).sum(1)                              # global count over its positions
    sizes = np.bincount(labels, minlength=Kc); nk = sizes[labels]
    within = (oc_in  - d) / np.maximum(nk - 1, 1) / np.maximum(d, 1)
    across = (oc_tot - oc_in) / np.maximum(N - nk, 1) / np.maximum(d, 1)
    p0 = float(occ.mean())                                    # chance coverage rate
    valid  = (d > 0) & (nk > 1) & (nk < N)
    ratio  = within[valid] / np.maximum(across[valid], p0)
    return float(ratio.mean()), int(valid.sum())

ratios = [seqlet_overlap_norm(lab)[0] for _, lab, _ in label_sets]
for nm, r, s in zip(names, ratios, survivors):
    print(f"{nm}  within/across overlap ratio={r:.2f}  ({s} clusters)")

fig, ax = plt.subplots(figsize=(0.6 * len(names) + 3, 5))
ax.bar(range(len(names)), ratios, color=colors)
ax.axhline(1.0, color="k", lw=1, ls="--")                    # 1 = no better than chance
for i, (r, s) in enumerate(zip(ratios, survivors)):
    ax.text(i, r + 0.02 * max(ratios), f"{r:.2f}x\n{s} cl", ha="center", va="bottom", fontsize=7)
ax.set_xticks(range(len(names))); ax.set_xticklabels(names, rotation=45, ha="right", fontsize=7)
ax.set_ylim(0, max(ratios) * 1.2)
ax.set_ylabel("seqlet-overlap enrichment  (within-cluster / across-cluster)")
plt.tight_layout(); plt.show()
```
